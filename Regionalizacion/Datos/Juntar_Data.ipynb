```python
import pandas as pd
import os
import glob

# ==========================================================
# 1. RUTA DE LA CARPETA
# ==========================================================
archivos = [
    r"C:\Users\Alex\Documents\Programación\PAP\Actualización de datos\2023_1.csv",
    r"C:\Users\Alex\Documents\Programación\PAP\Actualización de datos\2023_2.csv",
    r"C:\Users\Alex\Documents\Programación\PAP\Actualización de datos\2024_1.csv",
    r"C:\Users\Alex\Documents\Programación\PAP\Actualización de datos\2024_2.csv",
    r"C:\Users\Alex\Documents\Programación\PAP\Actualización de datos\2025_1.csv",
    r"C:\Users\Alex\Documents\Programación\PAP\Actualización de datos\2025_2.csv",
    r"C:\Users\Alex\Documents\Programación\PAP\Actualización de datos\2025_Enero.csv",
    r"C:\Users\Alex\Documents\Programación\PAP\Actualización de datos\2026_Enero.csv",
]

print(f"Se encontraron {len(archivos)} archivos.")

# ==========================================================
# 2. LEER Y UNIR
# ==========================================================
lista_df = []

for archivo in archivos:
    print(f"Leyendo: {os.path.basename(archivo)}")

    if archivo.endswith(".csv"):
        df = pd.read_csv(archivo, encoding="latin1")
    else:
        df = pd.read_excel(archivo)

    lista_df.append(df)

df_final = pd.concat(lista_df, ignore_index=True)

# ==========================================================
# 3. ASEGURAR TIPOS CORRECTOS
# ==========================================================
df_final['fecha'] = pd.to_datetime(df_final['fecha'])
df_final['sector_economico_1'] = df_final['sector_economico_1'].astype(int)
df_final['sector_economico_2'] = df_final['sector_economico_2'].astype(int)
df_final['cve_entidad'] = df_final['cve_entidad'].astype(int)

# ==========================================================
# 4. ORDENAR
# ==========================================================
df_final = df_final.sort_values(
    by=['sector_economico_1', 'sector_economico_2', 'fecha', 'cve_entidad']
).reset_index(drop=True)

print(f"\nArchivo final: {df_final.shape[0]:,} filas")

# ==========================================================
# 5. GUARDAR
# ==========================================================
df_final.to_csv("TA_2023-2026.csv", index=False)
# O mejor:
# df_final.to_parquet("panel_completo.parquet")
```
